```python
import numpy as np
import pandas as pd
from scipy import stats
from sklearn.linear_model import LinearRegression

import matplotlib.pyplot as plt
```

```python
# Params
plot = True # Plot the linear regression
path = './ms2-2_copy_number_estimation.xls'# Path to the original raw data
```

```python
def generate_axes(df):
    """Generate the x and y axes from a to feed a linear regression model"""
    x_axis = np.log10(np.logspace(0, -len(df)+1, num=len(df))).reshape(-1,1)
    y_axis = np.array(df['CT'])
    return (x_axis, y_axis)

def compute_lr(df, title, plot=plot):
    """Compute a linear regression model and return the fitted regresion model and the efficiency of the primers"""
    # Compute a linear regression model
    model = LinearRegression()
    axes = generate_axes(df)
    model.fit(axes[0], axes[1])
    
    # Retrieve & print useful values from the lr
    r_score = model.score(axes[0], axes[1])
    slope = float(model.coef_)
    y_origin = model.intercept_
    eff = -1+10**(-1/slope)
    
    print('r_score = {r_score}'.format(r_score=r_score))
    print('slope = {slope}'.format(slope=slope))
    print('y_origin = {y_origin}'.format(y_origin=y_origin))
    print('Efficiency = {eff}'.format(eff=eff))
    
    # Plot
    if plot:
        plt.close('All')
        plt.axline((0, y_origin), slope=slope, linewidth=2)
        plt.plot(axes[0], axes[1], 'o')
        plt.xlabel('Relative conc (log10)')
        plt.ylabel('Ct')
        plt.legend(['y = {slope:.3f}x + {y_origin:.3f}'.format(slope=slope, y_origin=y_origin)])
        plt.title(title)
    
    return [model, eff]
```

```python
def find_intercept(ct, model):
    """Find the intercept of the value Ct for a given linear regression model"""
    return (ct - model.intercept_) / model.coef_
```

```python
def compute_ratio(df, eff_her1=2, eff_ef1a=2):
    new_df = df.copy(deep=True)
    new_df['Ratio'] = eff_her1**(-df['CT_her1'])/eff_ef1a**(-df['CT_ef1a'])
    return new_df
```

```python
def propagate_uncertainties(ratio_df):
    # Separate groups: alternating rows, starting with wild-type (neg)
    neg_ratios = ratio_df['Ratio'][::2]
    pos_ratios = ratio_df['Ratio'][1::2]
    
    # Means
    avg_neg = neg_ratios.mean()
    avg_pos = pos_ratios.mean()
    
    # Standard errors
    sem_neg = stats.sem(neg_ratios)
    sem_pos = stats.sem(pos_ratios)
    
    # Fold change
    fold_change = avg_pos / avg_neg
    
    # Error propagation of fold change
    err_fc = fold_change * np.sqrt((sem_neg / avg_neg) ** 2 + (sem_pos / avg_pos) ** 2)
    
    # Copy number assuming that both wild-type and transgene have 2 endogenous copies of her1
    copy_number = 2 * fold_change - 2
    
    # Error propagation of copy number
    err_cn = 2 * err_fc
    
    return {
        "copy_number": copy_number,
        "error_copy_number": err_cn,
        "fold_change": fold_change,
        "error_fc": err_fc,
        "mean_neg": avg_neg,
        "mean_pos": avg_pos,
        "sem_neg": sem_neg,
        "sem_pos": sem_pos
    }
```

```python
df = pd.read_excel(path, sheet_name='Results', 
                   skiprows=range(0,46),
                   usecols=[0, 1, 2, 3, 4, 5, 8, 9 , 10],
                   na_values='Undetermined')
df = df.dropna(axis=0)
```

```python
# Split the samples used for building a calibration curve
df_standards_her1 = df.loc[(df['Task']=='STANDARD') & (df['Target Name'] == 'her1'),:]
df_standards_ef1a = df.loc[(df['Task']=='STANDARD') & (df['Target Name'] == 'Ef1a'),:]
```

```python
# Sort the dataframaes by Sample Name
df_standards_her1.sort_values(by='Sample Name', inplace=True)
df_standards_ef1a.sort_values(by='Sample Name', inplace=True)
```

```python
# Generate the mean and the SD for the calibrations
df_standards_her1_mean = df_standards_her1[['Sample Name', 'CT', 'Ct SD']].groupby('Sample Name').mean().reset_index()
df_standards_ef1a_mean = df_standards_ef1a[['Sample Name', 'CT', 'Ct SD']].groupby('Sample Name').mean().reset_index()
standard_mask = df_standards_her1_mean['Sample Name'].str.contains('NEG')
her1_EMB_neg = df_standards_her1_mean[standard_mask]
standard_mask = df_standards_ef1a_mean['Sample Name'].str.contains('NEG')
ef1a_EMB_neg = df_standards_ef1a_mean[standard_mask]
```

```python
cal_her1 = compute_lr(her1_EMB_neg[:-2], 'her1 calibration curve')
```

```python
cal_ef1a = compute_lr(ef1a_EMB_neg[:-1], 'ef1a calibration curve')
```

```python
# Split samples by primers
her1_sample = df.loc[(df['Target Name'] == 'her1') & (df['Task'] != 'STANDARD') & (df['Sample Name'] != 'H20'), :].reset_index()
ef1a_sample = df.loc[(df['Target Name'] == 'Ef1a') & (df['Task'] != 'STANDARD') & (df['Sample Name'] != 'H20'), :].reset_index()
her1_sample_mean = her1_sample[['CT', 'Ct SD']].groupby(her1_sample['Sample Name']).mean().reset_index()
ef1a_sample_mean = ef1a_sample[['CT', 'Ct SD']].groupby(ef1a_sample['Sample Name']).mean().reset_index()
```

```python
# Compute Coefficient of Variation (CV) as percentage
her1_sample_mean['CV'] = her1_sample_mean['Ct SD']*100/her1_sample_mean['CT']
ef1a_sample_mean['CV'] = ef1a_sample_mean['Ct SD']*100/ef1a_sample_mean['CT']
```

```python
# 1-embryo measurements 
# her1_1emb = her1_sample_mean.loc[(her1_sample_mean['Sample Name'].str.contains('pos1')) | (her1_sample_mean['Sample Name'].str.contains('neg')) & (~her1_sample_mean['Sample Name'].str.contains('neg3')), :]
# ef1a_1emb = ef1a_sample_mean.loc[(ef1a_sample_mean['Sample Name'].str.contains('pos1')) | (ef1a_sample_mean['Sample Name'].str.contains('neg')) & (~ef1a_sample_mean['Sample Name'].str.contains('neg3')), :]
```

```python
# 3-embryo measurements
her1_3emb = her1_sample_mean.loc[her1_sample_mean['Sample Name'].str.contains('pos3') | her1_sample_mean['Sample Name'].str.contains('neg3'), :]
ef1a_3emb = ef1a_sample_mean.loc[ef1a_sample_mean['Sample Name'].str.contains('pos3') | ef1a_sample_mean['Sample Name'].str.contains('neg3'), :]
```

```python
# Group CT measurements per sample & gene - use only 3-embryo measurements
total_CT = pd.concat([her1_3emb['Sample Name'], her1_3emb['CT']],axis=1)
total_CT.rename(columns={'CT':'CT_her1'}, inplace=True)
total_CT['CT_ef1a'] = ef1a_3emb['CT']
```

```python
ratios_df = compute_ratio(total_CT, cal_her1[1]+1,cal_ef1a[1]+1)
```

```python
ratios_df = ratios_df[['Sample Name', 'CT_her1', 'CT_ef1a', 'Ratio']]
```

```python
results_copy_number = propagate_uncertainties(ratios_df)
```
